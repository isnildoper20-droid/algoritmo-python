## 1. NumPy – Manipulación de arrays

```python
import numpy as np

matriz = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(matriz)
```

### Actividades: elemento central, segunda fila y tercera columna

```python
# Elemento central
print("Elemento central:", matriz[1, 1])

# Segunda fila
print("Segunda fila:", matriz[1, :])

# Tercera columna
print("Tercera columna:", matriz[:, 2])
```

## 2. Operaciones matemáticas

```python
datos = np.array([12, 18, 20, 25, 30, 35, 40])

print("Promedio:", np.mean(datos))
print("Máximo:", np.max(datos))
print("Mínimo:", np.min(datos))
print("Desviación:", np.std(datos))
```

## 3. Generación de datos aleatorios – Distribución normal

```python
import matplotlib.pyplot as plt

datos_normales = np.random.normal(
    loc=50,    # media
    scale=10,  # desviación estándar
    size=1000  # cantidad de datos
)

plt.hist(datos_normales, bins=30, color="skyblue", edgecolor="black")
plt.title("Distribución normal generada (loc=50, scale=10)")
plt.xlabel("Valor")
plt.ylabel("Frecuencia")
plt.show()
```

## 4. Matplotlib – Gráfico de líneas

> Concepto a investigar: **accuracy** (precisión del modelo) y **época** (una pasada completa del algoritmo de entrenamiento por todo el conjunto de datos). La explicación completa está en `README.md`.

```python
epocas = [1, 2, 3, 4, 5, 6, 7, 8]
accuracy = [60, 68, 75, 81, 85, 88, 90, 92]

plt.plot(epocas, accuracy, marker="o")
plt.title("Accuracy por época")
plt.xlabel("Épocas")
plt.ylabel("Accuracy")
plt.grid(True)
plt.show()
```

## 5. Histograma

```python
datos_hist = np.random.normal(70, 15, 1000)

plt.hist(datos_hist, bins=30, color="salmon", edgecolor="black")
plt.title("Histograma de datos (media=70, sd=15)")
plt.xlabel("Valor")
plt.ylabel("Frecuencia")
plt.show()
```

## 6. Gráfico de barras

```python
clases = ['Sano', 'Roya', 'Mosaico', 'Mancha']
cantidad = [1200, 850, 430, 320]

plt.bar(clases, cantidad, color=["green", "orange", "red", "brown"])
plt.title("Distribución de clases")
plt.xlabel("Clase")
plt.ylabel("Cantidad de muestras")
plt.show()
```

## 7. Seaborn – Mapa de calor (matriz de correlación)

```python
import pandas as pd
import seaborn as sns

datos_df = pd.DataFrame({
    'Edad': [20, 25, 30, 35, 40],
    'Horas': [2, 4, 6, 8, 10],
    'Nota': [60, 70, 80, 90, 100]
})

correlacion = datos_df.corr()

sns.heatmap(correlacion, annot=True, cmap="coolwarm")
plt.title("Matriz de correlación")
plt.show()
```

## 8. Boxplot

```python
sns.boxplot(x=datos_df["Nota"])
plt.title("Boxplot de la variable Nota")
plt.show()
```
